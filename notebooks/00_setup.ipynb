# 00 - Configuração do ambiente

Este notebook prepara a estrutura utilizada pelo projeto no Databricks.

São criados os schemas correspondentes às camadas Bronze, Silver e Gold, além do volume utilizado para armazenar os arquivos brutos das fontes de dados.

A organização segue uma arquitetura em camadas:

- **RAW:** arquivos originais utilizados como fonte;
- **Bronze:** dados ingeridos com pouca transformação, preservando a estrutura original;
- **Silver:** dados tratados, padronizados e preparados para análise;
- **Gold:** tabelas analíticas integradas e agregadas, utilizadas nas análises finais.

Também são criadas pastas específicas para os arquivos do Brent, PPI e preços de combustíveis da ANP.

```python
%sql

CREATE SCHEMA IF NOT EXISTS workspace.raw;
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS workspace.bronze;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.silver;
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS workspace.gold;
```
